```python
print("Consumption Function Calculator")

income = int(input("Enter Income: "))
mpc = float(input("Enter MPC (between 0 and 1): "))

consumption = mpc * income
savings = income - consumption

print("Consumption Expenditure:", consumption)
print("Savings:", savings)
```
