```python
# Cost of Living Comparison Tool


city_a_expense = int(input("Enter monthly expenses for City A: "))
city_b_expense = int(input("Enter monthly expenses for City B: "))

# Comparing expenses
difference = abs(city_a_expense - city_b_expense)

if city_a_expense > city_b_expense:
    print("City A has a higher cost of living.")
    print("Difference =", round(difference, 2))
elif city_b_expense > city_a_expense:
    print("City B has a higher cost of living.")
    print("Difference =", round(difference, 2))
else:
    print("Both cities have the same cost of living.")
```
