```python
print("Budget Deficit/Surplus Calculator")

total_revenue = int(input("Enter Total Revenue: "))
total_expenditure = int(input("Enter Total Expenditure: "))

# Calculating difference
difference = total_revenue - total_expenditure

print("Budget Difference:", difference)

# Classifying budget
if difference > 0:
    print("The government has a Budget Surplus.")
elif difference < 0:
    print("The government has a Budget Deficit.")
else:
    print("The budget is Balanced.")
```
