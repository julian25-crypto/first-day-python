```python
print("Population Growth Rate Calculator")

population_old = int(input("Enter Population of Previous Year: "))
population_new = int(input("Enter Population of Current Year: "))

# Calculate growth rate
growth_rate = ((population_new - population_old) / population_old) * 100

print("Population Growth Rate:", growth_rate, "%")

# Check increase or decrease
if growth_rate > 0:
    print("Population is Increasing.")
elif growth_rate < 0:
    print("Population is Decreasing.")
else:
    print("No Change in Population.")
```
