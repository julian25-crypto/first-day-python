```python
# Household Monthly Budget Analyzer

# Taking input from user
income = int(input("Enter total monthly income: "))

rent = int(input("Enter monthly rent expense: "))
food = int(input("Enter monthly food expense: "))
transport = int(input("Enter monthly transport expense: "))
utilities = int(input("Enter monthly utilities expense: "))
others = int(input("Enter other monthly expenses: "))

# Calculating total expenditure
total_expense = rent + food + transport + utilities + others

# Calculating surplus or deficit
balance = income - total_expense

print(" Monthly Budget Summary")
print("Total Income: ", income)
print("Total Expenditure: ", total_expense)

if balance > 0:
    print("Status: Surplus (Savings)")
    print("Amount Saved: ", balance)
elif balance < 0:
    print("Status: Deficit")
    print("Amount Over Budget: ", abs(balance))
else:
    print("Status: Balanced Budget (No Savings, No Loss)")
```
