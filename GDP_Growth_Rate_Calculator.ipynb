```python
# GDP Growth Rate Calculator

# Taking input from user
gdp_previous = int(input("Enter GDP of previous year: "))
gdp_current = int(input("Enter GDP of current year: "))

# Calculating GDP growth rate
growth_rate = ((gdp_current - gdp_previous) / gdp_previous) * 100

print("GDP Growth Analysis")
print("Previous Year GDP:", gdp_previous)
print("Current Year GDP:", gdp_current)
print("Growth Rate: {:.2f}%".format(growth_rate))

# Checking economic condition
if growth_rate > 0:
    print("The economy is growing.")
elif growth_rate < 0:
    print("The economy is shrinking.")
else:
    print("No economic growth.")
```
