```python
# Inflation Rate Calculator

# Taking input from user
previous_price = int(input("Enter the price of the commodity last year: "))
current_price = int(input("Enter the price of the commodity this year: "))

# Calculating inflation rate
inflation_rate = ((current_price - previous_price) / previous_price) * 100

print("Inflation Analysis")
print("Previous Year Price:", previous_price)
print("Current Year Price:", current_price)
print("Inflation Rate: {:.2f}%".format(inflation_rate))

# Checking price movement
if inflation_rate > 0:
    print("Prices have increased (Inflation).")
elif inflation_rate < 0:
    print("Prices have decreased (Deflation).")
else:
    print("No change in prices.")
```
