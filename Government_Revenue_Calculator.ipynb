```python
print("Government Revenue Calculator")

# Taking inputs
tax = float(input("Enter revenue from Taxes: "))
fees = float(input("Enter revenue from Fees: "))
fines = float(input("Enter revenue from Fines: "))

# Calculating total revenue
total_revenue = tax + fees + fines

print("Total Government Revenue:", total_revenue)

# Finding highest revenue source
if tax >= fees and tax >= fines:
    print("Taxes contribute the highest revenue.")
elif fees >= tax and fees >= fines:
    print("Fees contribute the highest revenue.")
else:
    print("Fines contribute the highest revenue.")
```
