```python
# Price Elasticity Classifier (Extended Version)

# Taking inputs
old_price = float(input("Enter old price: "))
new_price = float(input("Enter new price: "))
old_demand = float(input("Enter old demand: "))
new_demand = float(input("Enter new demand: "))

# Calculating percentage changes
percentage_change_demand = (new_demand - old_demand) / old_demand
percentage_change_price = (new_price - old_price) / old_price

# Handling special cases
if percentage_change_price == 0:
    elasticity = float('inf')   # Price unchanged
else:
    elasticity = percentage_change_demand / percentage_change_price

elasticity_abs = abs(elasticity)

print("Price Elasticity of Demand =", round(elasticity_abs, 2))

# Classification
if percentage_change_demand == 0:
    print("Demand is PERFECTLY INELASTIC")
elif percentage_change_price == 0 and percentage_change_demand != 0:
    print("Demand is PERFECTLY ELASTIC")
elif elasticity_abs > 1:
    print("Demand is ELASTIC")
elif elasticity_abs < 1:
    print("Demand is INELASTIC")
elif elasticity_abs == 1:
    print("Demand is UNITARY ELASTIC")
```
