```python
# Unemployment Rate Calculator

# Taking input from user
labor_force = float(input("Enter total labor force: "))
unemployed = float(input("Enter number of unemployed people: "))

# Calculating unemployment rate
unemployment_rate = (unemployed / labor_force) * 100

print("Unemployment Analysis")
print("Total Labor Force:", labor_force)
print("Number of Unemployed:", unemployed)
print("Unemployment Rate: {:.2f}%".format(unemployment_rate))
```
