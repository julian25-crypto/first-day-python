```python
# Farmer Income Estimator

# Taking input from user
quantity = float(input("Enter quantity of crop produced: "))
price_per_unit = float(input("Enter price per unit: "))
total_cost = float(input("Enter total cost of production: "))

# Calculating total revenue
total_revenue = quantity * price_per_unit

# Calculating profit or loss
profit_loss = total_revenue - total_cost

print("\n------ Farmer Income Analysis ------")
print("Total Revenue:", total_revenue)
print("Total Cost of Production:", total_cost)

if profit_loss > 0:
    print("Profit:", round(profit_loss, 2))
elif profit_loss < 0:
    print("Loss:", round(abs(profit_loss), 2))
else:
    print("No Profit, No Loss (Break-Even)")

```
