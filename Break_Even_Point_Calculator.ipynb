```python
# Break-Even Point Calculator

# Taking input from user
fixed_cost = float(input("Enter fixed cost: "))
variable_cost = float(input("Enter variable cost per unit: "))
selling_price = float(input("Enter selling price per unit: "))

# Checking if calculation is possible
if selling_price <= variable_cost:
    print("\nBreak-even is not possible because selling price is less than or equal to variable cost.")
else:
    break_even_quantity = fixed_cost / (selling_price - variable_cost)

    print("Break-Even Analysis")
    print("Fixed Cost:", fixed_cost)
    print("Variable Cost per Unit:", variable_cost)
    print("Selling Price per Unit:", selling_price)
    print("Break-Even Quantity (units):", round(break_even_quantity, 2))
```
