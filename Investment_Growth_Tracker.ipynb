```python
print("Investment Growth Tracker")

initial_investment = float(input("Enter Initial Investment: "))
growth_rate = float(input("Enter Annual Growth Rate (in %): "))
years = int(input("Enter Number of Years: "))

# Convert percentage to decimal
rate = growth_rate / 100

# Compound growth formula
final_value = initial_investment * (1 + rate) ** years

print("Investment Value after", years, "years:", final_value)
```
