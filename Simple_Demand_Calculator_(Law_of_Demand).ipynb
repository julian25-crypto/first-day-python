```python
# Simple Demand Calculator (Law of Demand)

# Taking input from user
base_price = int(input("Enter the base price of the product: "))
base_demand = int(input("Enter the base demand at base price: "))

new_price = int(input("Enter the new price of the product: "))

# Using simple inverse relationship (Demand × Price = Constant)
constant = base_price * base_demand
new_demand = constant / new_price

print("Demand Analysis")
print("Base Price:", base_price)
print("Base Demand:", base_demand)
print("New Price:", new_price)
print("New Calculated Demand:", round(new_demand, 2))

# Checking Law of Demand
if new_price > base_price:
    print("Price increased → Demand decreased (Law of Demand)")
elif new_price < base_price:
    print("Price decreased → Demand increased (Law of Demand)")
else:
    print("Price unchanged → Demand remains the same")
```
