```python
print("Cost Structure Analyzer")

fixed_cost = int(input("Enter Fixed Cost: "))
variable_cost_per_unit = int(input("Enter Variable Cost per Unit: "))
output = int(input("Enter Output Level (number of units produced): "))

# Calculating total cost
total_cost = fixed_cost + (variable_cost_per_unit * output)

# Calculating cost per unit
cost_per_unit = total_cost / output

# Display results
print("Total Cost:", total_cost)
print("Cost per Unit:", cost_per_unit)
```
