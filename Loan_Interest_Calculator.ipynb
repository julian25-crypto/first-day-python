```python
# Loan Interest Calculator

# Taking inputs
principal = int(input("Enter principal amount: "))
rate = int(input("Enter rate of interest (in % per year): "))
time = int(input("Enter time period (in years): "))

# Calculating Simple Interest
interest = (principal * rate * time) / 100

# Calculating Total Amount to be Repaid
total_amount = principal + interest

# Displaying results
print("\nTotal Interest =", round(interest, 2))
print("Total Repayment Amount =", round(total_amount, 2))
```
