```python
print("Income Classification System")

annual_income = int(input("Enter Annual Income: "))

# Classifying income
if annual_income < 30000:
    print("Income Group: Low Income")
elif annual_income >= 30000 and annual_income <= 100000:
    print("Income Group: Middle Income")
else:
    print("Income Group: High Income")
```
