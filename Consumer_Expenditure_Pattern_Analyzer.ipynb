```python
# Consumer Expenditure Pattern Analyzer

# Taking input from user
food = float(input("Enter expenditure on Food: "))
housing = float(input("Enter expenditure on Housing: "))
education = float(input("Enter expenditure on Education: "))
entertainment = float(input("Enter expenditure on Entertainment: "))

# Calculating total expenditure
total = food + housing + education + entertainment

# Calculating percentage share
food_percent = (food / total) * 100
housing_percent = (housing / total) * 100
education_percent = (education / total) * 100
entertainment_percent = (entertainment / total) * 100

# Finding highest spending category
expenses = {
    "Food": food,
    "Housing": housing,
    "Education": education,
    "Entertainment": entertainment
}

highest_category = max(expenses, key=expenses.get)

print("Expenditure Analysis")
print("Total Expenditure:", total)

print("\nPercentage Share:")
print("Food: {:.2f}%".format(food_percent))
print("Housing: {:.2f}%".format(housing_percent))
print("Education: {:.2f}%".format(education_percent))
print("Entertainment: {:.2f}%".format(entertainment_percent))

print("\nHighest Spending Category:", highest_category)
```
