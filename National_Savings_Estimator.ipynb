```python
print("National Savings Estimator")


household_savings = int(input("Enter Household Savings: "))
corporate_savings = int(input("Enter Corporate Savings: "))
government_savings = int(input("Enter Government Savings: "))

# Calculating total national savings
total_savings = household_savings + corporate_savings + government_savings

print("Total National Savings:", total_savings)

# Determining if savings are sufficient
# (Assuming sufficient savings if total is greater than 10,000)
if total_savings >= 10000:
    print("The country is saving sufficiently.")
else:
    print("The country is NOT saving sufficiently.")
```
