```python
print("Market Share Calculator")

company_sales = int(input("Enter Company Sales: "))
total_market_sales = int(input("Enter Total Market Sales: "))

market_share = (company_sales / total_market_sales) * 100

print("Market Share:", market_share, "%")

# Classify market position
if market_share < 20:
    print("Small Market Share")
elif market_share <= 50:
    print("Medium Market Share")
else:
    print("Dominant Market Share")
```
