```python
# Profit and Loss Analyzer

# Taking input from user
revenue = float(input("Enter total revenue: "))
cost = float(input("Enter total cost: "))

# Calculating profit or loss
result = revenue - cost

print("\n------ Profit and Loss Analysis ------")
print("Total Revenue:", revenue)
print("Total Cost:", cost)

if result > 0:
    print("Profit Amount:", round(result, 2))
elif result < 0:
    print("Loss Amount:", round(abs(result), 2))
else:
    print("No Profit, No Loss (Break-Even)")
```
