```python
print("Market Supply Calculator")



old_price = int(input("Enter the old price of the product: "))
new_price = int(input("Enter the new price of the product: "))
base_supply = int(input("Enter the base supply quantity: "))

# Direct relationship rule (Supply changes proportionally with price)
new_supply = base_supply * (new_price / old_price)

# Display result
print("New Supply Quantity:", new_supply)

# Checking increase or decrease
if new_price > old_price:
    print("Price increased → Supply increased (Law of Supply)")
elif new_price < old_price:
    print("Price decreased → Supply decreased (Law of Supply)")
else:
    print("Price did not change → Supply remains the same")
```
